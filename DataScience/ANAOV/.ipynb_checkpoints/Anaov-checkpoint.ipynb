```python
#Oneway ANAVO
import pandas as pd
dataset=pd.read_csv("placement.csv")
```

```python
dataset
dataset=dataset.dropna()
dataset
```

```python
#One-way Anaov
import scipy.stats as stats
f_value,p_value=stats.f_oneway(dataset["hsc_p"],dataset["ssc_p"])
sign_value=0.05
print(f_value)
print(p_value)
if p_value<sign_value:
    print("Reject H0: There is a significant difference  between the hSC pass & SSC pass.")
else:
    print("Fail to reject H0: We cannot conclude a significant difference based on this sample.")
```

```python
dataset=pd.read_csv("PreprocessChronicKidney.csv")
dataset
```

```python
#Two-way Anaov

import statsmodels.api as sm
from statsmodels.formula.api import ols
df = pd.DataFrame(dataset)
df
model = ols('age ~ C(bp) + C(sg) + C(bp):C(sg)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
```
